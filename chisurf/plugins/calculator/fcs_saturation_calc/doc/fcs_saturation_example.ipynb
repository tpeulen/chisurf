# FCS Saturation Calculator Notebook Example

This notebook demonstrates how to use the `fcs_saturation_calc` plugin API to calculate non-Gaussian FCS volume saturation for standard dyes and complex 6-state kinetics (Cy5 cis/trans).

```python
import matplotlib.pyplot as plt

from chisurf.plugins.calculator.fcs_saturation_calc import simulate_cy5_saturation

# 1. Simulate Cy5 6-state FCS saturation
tau_ms, g_unpert, g_sat = simulate_cy5_saturation(
    power_mW=3.0, extinction=250000.0, w_r_nm=250.0, w_z_nm=1000.0, D_um2s=10.0
)

print(f"Unperturbed G(0): {g_unpert[0]:.4f}")
print(f"Saturated G(0):   {g_sat[0]:.4f}")
print(f"Amplitude Ratio:  {g_sat[0] / g_unpert[0]:.4f}")
```

```python
# 2. Plot FCS curves
plt.figure(figsize=(7, 4))
plt.semilogx(tau_ms, g_unpert, "b--", label="Unperturbed (Gaussian)")
plt.semilogx(tau_ms, g_sat, "r-", label="Saturated (Cy5 6-state)")
plt.xlabel("$\tau$ (ms)")
plt.ylabel("$G(\tau)$")
plt.title("FCS Saturation Effect on Cy5 Kinetics")
plt.grid(True, which="both", ls=":")
plt.legend()
plt.tight_layout()
plt.show()
```
